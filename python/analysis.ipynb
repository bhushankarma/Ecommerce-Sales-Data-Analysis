```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r'Superstore.csv',encoding="latin1")
df
```

```python
df.head()
df.shape
df.info()
df.describe()
```

```python
df.duplicated().sum()
df.nunique()
df['Customer ID'].nunique()
df['Order Date'].nunique()
```

```python
df['Order Date'] = pd.to_datetime(df['Order Date'],dayfirst=True)
df['Ship Date'] = pd.to_datetime(df['Ship Date'],dayfirst=True)
df.info()
```

```python
df['Order Date'].min()
df['Order Date'].max()
df['Ship Date'].min()
df['Ship Date'].max()
```

```python
df['Shipping Days'] = (df['Ship Date'] - df['Order Date']).dt.days
df['Shipping Days'].describe()
df['Shipping Days'].value_counts().sort_index()
```

```python
df['Sales'].min()
df['Sales'].max()
(df['Sales'] <= 0).sum()

df['Quantity'].min()
df['Quantity'].max()
(df['Quantity'] <= 0).sum()

df['Discount'].min()
df['Discount'].max()
(df['Discount'] < 0).sum()
(df['Discount'] > 1).sum()

df['Profit'].min()
df['Profit'].max()
(df['Profit'] < 0).sum()

(df['Shipping Days'] < 0).sum()
```

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
df['Quarter'] = df['Order Date'].dt.quarter

df[['Order Date','Year','Month','Quarter']].head(10)

df['Order Date'].dt.month_name()
df['Month Name'] =df['Order Date'].dt.month_name()
df[['Order Date','Month','Month Name','Quarter']].head(10)
```

```python
df.groupby('Month')['Sales'].sum()

monthly_sales = df.groupby(['Year','Month'])['Sales'].sum()
monthly_sales

yearly_sales = df.groupby('Year')['Sales'].sum()
yearly_sales

category_sales = df.groupby('Category')['Sales'].sum()
category_sales

category_profit = df.groupby('Category')['Profit'].sum()
category_profit
```

```python
category_margin = (category_profit / category_sales) * 100
category_margin

subcategory_sales = df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending=False)
subcategory_sales

subcategory_profit = df.groupby('Sub-Category')['Profit'].sum().sort_values(ascending=False)
subcategory_profit

subcategory_margin = (subcategory_profit / subcategory_sales) * 100
subcategory_margin.sort_values()
```

```python
df['Discount'].describe()

discount_profit = df.groupby('Discount')['Profit'].agg(['sum','mean'])
discount_profit

tables_discount = df[df['Sub-Category'] == 'Tables'].groupby('Discount')['Profit'].agg(['sum','mean'])
tables_discount

total_products = df[df['Sub-Category'] == 'Tables'].groupby('Product Name')['Profit'].sum().sort_values()
total_products
```

```python
tables_analysis = df[df['Sub-Category'] == 'Tables'].groupby('Discount').agg({
    'Sales':'sum',
    'Profit':'sum',
    'Quantity':'sum'
})
tables_analysis

tables_analysis['Profit Margin %'] = (tables_analysis['Profit'] / tables_analysis['Sales']) * 100
tables_analysis
```

```python
tables_products = df[df['Sub-Category'] == 'Tables'].groupby('Product Name').agg({
    'Sales':'sum',
    'Profit':'sum',
    'Quantity':'sum'
}).sort_values('Profit')

tables_products.head(10)

tables_products['Profit Margin %'] = (tables_products['Profit'] / tables_products['Sales']) * 100
tables_products.head(10)
```

```python
customer_orders = df.groupby('Customer ID')['Order ID'].nunique()
customer_orders.head(10)

customer_orders.value_counts().sort_index()

one_time_customers = (customer_orders == 1).sum()
repeat_customers = (customer_orders > 1).sum()
one_time_customers, repeat_customers

one_time_percentage = (one_time_customers / len(customer_orders)) * 100
repeat_percentage = (repeat_customers / len(customer_orders)) * 100
one_time_percentage, repeat_percentage

customer_sales = df.groupby('Customer ID').agg({
    'Sales':'sum',
    'Profit':'sum',
    'Order ID':'nunique'
}).sort_values('Sales',ascending=False)
customer_sales.head(10)

customer_profit = customer_sales.sort_values('Profit',ascending=False)
customer_profit.head(10)

customer_sales['Profit Margin %'] = (customer_sales['Profit'] / customer_sales['Sales']) * 100
customer_sales.head(10)

customer_margin = customer_sales.sort_values('Profit Margin %',ascending=False)
customer_margin.head(10)
```

```python
customer_rfm = df.groupby('Customer ID').agg({
    'Order Date':'max',
    'Order ID':'nunique',
    'Sales':'sum'
})
customer_rfm.head(10)

snapshot_date = df['Order Date'].max()
customer_rfm['Recency'] = (snapshot_date - customer_rfm['Order Date']).dt.days
customer_rfm.head(10)


customer_rfm[['Recency','Order ID','Sales']].describe()
```

```python
customer_rfm['R_Score'] = pd.qcut(customer_rfm['Recency'].rank(method='first'),5,
                                  labels=[5,4,3,2,1]).astype(int)

customer_rfm['F_Score'] = pd.qcut(customer_rfm['Order ID'].rank(method='first'),5,
                                  labels=[1,2,3,4,5]).astype(int)

customer_rfm['M_Score'] = pd.qcut(customer_rfm['Sales'].rank(method='first'),5,
                                  labels=[1,2,3,4,5]).astype(int)

customer_rfm[['Recency','Order ID','Sales','R_Score','F_Score','M_Score']]


customer_rfm['RFM_Score'] = (customer_rfm['R_Score'] + customer_rfm['F_Score'] + customer_rfm['M_Score'])

customer_rfm[['Recency','Order ID','Sales','R_Score','F_Score','M_Score','RFM_Score']].head(10)

customer_rfm['RFM_Score'].value_counts().sort_index()


customer_rfm.groupby('RFM_Score')[['Recency','Order ID','Sales']].mean().round(2)
```

```python

def customer_segment(row):
    if row['RFM_Score'] >= 13:
        return 'Champions'
    elif row['RFM_Score'] >= 10:
        return 'Loyal Customer'
    elif row['RFM_Score'] >= 8:
        return 'Potential Loyalists'
    elif row['RFM_Score'] >= 6:
        return 'Need Attention'
    elif row['RFM_Score'] >= 4:
        return 'At Risk'
    else:
        return 'Lost Customers'
    
customer_rfm['Customer_Segment'] = customer_rfm.apply(customer_segment, axis=1)
customer_rfm['Customer_Segment'].value_counts()

```

```python
segment_analysis = customer_rfm.groupby('Customer_Segment').agg({
    'Sales':'sum',
    'Order ID':'sum',
    'Recency':'mean'
}).round(2)

segment_analysis


customer_profit = df.groupby('Customer ID')['Profit'].sum()
customer_profit

customer_rfm['Profit'] = customer_rfm.index.map(customer_profit)

segment_analysis = customer_rfm.groupby('Customer_Segment').agg({
    'Sales':'sum',
    'Profit':'sum',
    'Order ID':'sum',
    'Recency':'mean'
}).round(2)

segment_analysis

segment_analysis['Profit Margin %'] = (segment_analysis['Profit'] / segment_analysis['Sales']) * 100
segment_analysis.round(2)
```

```python
total_sales = df['Sales'].sum()
total_profit = df['Profit'].sum()
total_orders = df['Order ID'].nunique()
total_customers = df['Customer ID'].nunique()
total_quantity = df['Quantity'].sum()

print('Total Sales',total_sales)
print('Total Profit',total_profit)
print('Total Orders',total_orders)
print('Total Customers',total_customers)
print('Total Quantity',total_quantity)

overall_profit_margin = (total_profit / total_sales) * 100
aov = total_sales / total_orders

print('Overall Profit Margin',overall_profit_margin)
print('Average Order Value',aov)
```

```python
region_analysis = df.groupby('Region').agg({
    'Sales':'sum',
    'Profit':'sum',
    'Order ID':'nunique',
    'Customer ID':'nunique'
}).round(2)
region_analysis


region_analysis['Profit Margin %'] = (region_analysis['Profit'] / region_analysis['Sales']) * 100
region_analysis.round(2)


state_analysis = df.groupby('State').agg({
    'Sales':'sum',
    'Profit':'sum',
    'Order ID':'nunique',
    'Customer ID':'nunique'
})

state_analysis = state_analysis.sort_values('Sales',ascending=False)
state_analysis.head(10).round(2)

state_analysis['Profit Margin %'] = (state_analysis['Profit'] / state_analysis['Sales']) * 100
state_analysis.head(10).round(2)
```
